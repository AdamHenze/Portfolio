```python
#Question 5
array1 = np.array([1,3,3,4,5,6,7,8,9])
array2 = np.array([[1,2,3],[4,6,6]])
array3 = np.arange(1,11).reshape(5,2)

#function for finding median
def npmedian(array):
    newarr = array.flatten('F')
    size = len(newarr)
    #check for even or odd length, return middle
    if ((len(newarr) % 2) == 0):
        val = ((newarr[(len(newarr)//2) - 1] + newarr[(len(newarr)//2)]) / 2)
        return val
    else:
        val = newarr[(len(newarr)//2)]
        return val

#mode func
def npmode(array):
    newarr = array.flatten('F')
    size = len(newarr)
    if len(newarr) == 1:
        return newarr[0]

    #count members
    counts = dict(Counter(newarr))

    
    #if count goes above 1 or indx set new mode, return
    indx = 1
    for val in counts:
        if counts[val] > indx:
            indx = counts[val]
            mode = val
            return mode
    
print(npmedian(array1))
print(npmedian(array2))
print(npmedian(array3))

print(npmode(array1))
print(npmode(array2))
print(npmode(array3))
```

```python
#Question 8
#data
temperature_data = {
    'Maxine': [70.2, 72.5, 68.9],
    'James': [68.4, 71.2, 67.7],
    'Amanda': [75.1, 76.3, 74.8]
}
#make frame
temperatures = pd.DataFrame(temperature_data)
print("DataFrame (a):")
print(temperatures)

#set indices to times.
temperatures.index = ['Morning', 'Afternoon', 'Evening']
print("\nDataFrame (b):")
print(temperatures)

# c
maxine_temperatures = temperatures['Maxine']
print("\nColumn 'Maxine' Temperatures:")
print(maxine_temperatures)

#use loc to find data
# d
morning_temperatures = temperatures.loc['Morning']
print("\nRow 'Morning' Temperatures:")
print(morning_temperatures)

# e
morning_evening_temperatures = temperatures.loc[['Morning', 'Evening']]
print("\nRows 'Morning' and 'Evening' Temperatures:")
print(morning_evening_temperatures)

# f
amanda_maxine_temperatures = temperatures[['Amanda', 'Maxine']]
print("\nColumns 'Amanda' and 'Maxine' Temperatures:")
print(amanda_maxine_temperatures)

#use column and indice selection
# g
amanda_maxine_morning_afternoon = temperatures.loc[['Morning', 'Afternoon'], ['Amanda', 'Maxine']]
print("\nElements for 'Amanda' and 'Maxine' in 'Morning' and 'Afternoon':")
print(amanda_maxine_morning_afternoon)

#stats via describe
# h
temperature_statistics = temperatures.describe()
print("\nDescriptive Statistics:")
print(temperature_statistics)

# i
temperatures_transposed = temperatures.T
print("\nTransposed DataFrame:")
print(temperatures_transposed)

#sort based on axis
# j
temperatures_sorted = temperatures.sort_index(axis=1)
print("\nDataFrame with Sorted Column Names:")
print(temperatures_sorted)
```
