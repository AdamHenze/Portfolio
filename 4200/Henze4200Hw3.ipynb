In python a list is like an array however it can be dynamically sized, lists in python are mutable. A tuple in python is like a list however it stores objects as comma seperated values, however a tuple is immutable in python. A set is an unordered list of unique values and is mutable like a list. A dictionary is like a set in that it only stores unique values, however it stores data as key-value pairs. A dictionary is mutable but all keys need to be unique.

```python
from collections import Counter

sentences = []
for i in range(d + 1):
    sentence = input(f"Enter sentence {i + 1}: ")
    sentence = sentence.strip().lower()
    sentence = sentence.split()
    sentences.extend((sentence))

print(sentences)
word_counts = dict(Counter(sentences))
word_counts = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

i = 0
for i in range(len(word_counts)):
    for item in word_counts[i]:
        print(f"{item}\t", end="")
    print("")
```

```python
#3
def phonebook():
    contacts = {}  # Create an empty dictionary to store contacts

    while len(contacts) < m + 1:
        choice = input("Enter your choice ('add' to add a new contact or 'finish' to finish): ")

        if choice == 'add':
            last_name = input("Enter the last name of your friend: ")
            phone_digits = input("Enter the last four digits of their phone number: ")

            # Add the contact to the dictionary
            contacts[last_name] = phone_digits
            print(f"Contact for {last_name} added successfully!")

        elif choice == 'finish':
            break  # Exit the loop if the user chooses to finish

        else:
            print("Invalid choice. Please enter 'add' or 'finish'.")

    sorted_contacts_desc = dict(sorted(contacts.items(), key=lambda x: x[0], reverse=True))
    print("Contacts sorted by last name in descending order:")
    for last_name, last_four_digits in sorted_contacts_desc.items():
        print(f"{last_name}: {last_four_digits}")

    # Sort the dictionary in ascending order of values and print
    sorted_contacts_asc = dict(sorted(contacts.items(), key=lambda x: x[1]))
    print("\nContacts sorted by last four digits in ascending order:")
    for last_name, last_four_digits in sorted_contacts_asc.items():
        print(f"{last_name}: {last_four_digits}")
        
# Call the phonebook function to start adding contacts
phonebook()
```

```python
#4
import matplotlib.pyplot as plt
import numpy as np

# Define the list of responses
responses = [1, 2, 4, 4, 3, 5, 2, 1, 3, 3, 1, 4, 3, 3, 3, 2, 3, 3, 2]

# Calculate response frequencies
response_counts = {}
for response in responses:
    if response in response_counts:
        response_counts[response] += 1
    else:
        response_counts[response] = 1

# Sort the responses and frequencies
sorted_responses = sorted(response_counts.keys())
frequencies = [response_counts[response] for response in sorted_responses]

# Calculate percentages
total_responses = len(responses)
percentages = [(count / total_responses) * 100 for count in frequencies]

# Create a bar chart
plt.figure(figsize=(10, 6))
plt.bar(sorted_responses, percentages, align='center', alpha=0.7)
plt.xlabel('Response')
plt.ylabel('Percentage (%)')
plt.title('Response Frequencies and Percentages')
plt.xticks(sorted_responses)
plt.ylim(0, 100)

# Show the chart
plt.show()

# Calculate statistics
mean = np.mean(responses)
median = np.median(responses)
mode = np.argmax(np.bincount(responses))
minimum = min(responses)
maximum = max(responses)
range_val = maximum - minimum
variance = np.var(responses)
std_deviation = np.std(responses)

# Display statistics
print(f"Minimum: {minimum}")
print(f"Maximum: {maximum}")
print(f"Range: {range_val}")
print(f"Mean: {mean:.2f}")
print(f"Median: {median}")
print(f"Mode: {mode}")
print(f"Variance: {variance:.2f}")
print(f"Standard Deviation: {std_deviation:.2f}")
```
