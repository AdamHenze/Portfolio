```python

#question 2
#store words
words = []

#strip input store strings
paragraph = input()
paragraph = paragraph.strip().lower()
paragraph = paragraph.split()
words.extend((paragraph))

#function parses for letter counts
def letter_counter(words):
    letters = []
    for word in words:
        for letter in word:
            letters.extend(letter)
    letter_counts = dict(Counter(letters))
    return letter_counts

# count words and print
word_counts = dict(Counter(words))
print(word_counts)

#set keys
keys = list(word_counts.keys())
#parse keys in word counts prints data
for key in word_counts:
        print(f"{key} occurs {word_counts[key]} times")
#print letter data
print(letter_counter(words))
```

```python
# I was unable to complete question 3, apologies.

import matplotlib.pyplot as plt
import numpy as np
import random
import seaborn as sns

rolls1 = [random.randrange(1,7) for x in range(10 * m)]
rolls2 = [random.randrange(1,7) for x in range(10 * m)]

values1, frequencies1 = np.unique(rolls1, return_counts=True)
values2, frequencies2 = np.unique(rolls2, return_counts=True)

totalvalues = values1
totalfrequencies = frequencies1 + frequencies2
print(totalvalues,totalfrequencies)

title = f'Rolling Two 6 Sided Die {len(rolls1):,} Times'
sns.set_style('whitegrid')
axes = sns.barplot(x=totalvalues, y=totalfrequencies, palette='bright')

axes.set_title(title)
plt.Text(0.5,1,'XXXYYYZZZ')
axes.set(xlabel='Die Values', ylabel='Frequencies')
[plt.Text(92.667,0.5,'YYY'), plt.Text(0.5,58.7667,'XXX')]

axes.set_ylim(top=max(totalfrequencies) * 1.10)
for bar, frequency in zip(axes.patches, totalfrequencies):
    text_x = bar.get_x() + bar.get_width() / 2.0
    text_y = bar.get_height()
    text = f'{frequency:,}\n{frequency / len(rolls1):.3%}'
    axes.text(text_x, text_y, text, fontsize=11, ha='center', va='bottom')
```

```python
#question 4

# setup lists
word = ['big', 'small', 'good', 'bad', 'car']
synonyms = [
    ['large', 'huge', 'massive'],
    ['tiny', 'itsy', 'little'],
    ['nice', 'awesome', 'cool'],
    ['lame', 'shady', 'awful'],
    ['vehicle', 'automobile', 'carraige']
]
dictionary = {}

#create dict
for i in range(len(word)):
    dictionary.update({f"{word[i]}" : [synonyms[i]]})

#print tabular
for key in dictionary:
    print(key)
    for value in dictionary[key]:
        for item in value:
            print(f"\t{item}")
```
